```python
class NewsDataset(Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = list(labels)

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return {
            "input_ids": self.encodings["input_ids"][idx],
            "attention_mask": self.encodings["attention_mask"][idx],
            "labels": int(self.labels[idx])
        }


train_dataset = NewsDataset(
    train_encodings,
    train_part["label"]
)

val_dataset = NewsDataset(
    val_encodings,
    val_part["label"]
)

print("Train dataset:", len(train_dataset))
print("Validation dataset:", len(val_dataset))

print("\nПервый пример:")
print(train_dataset[0])
```

```python
def mean_pooling(last_hidden_state, attention_mask):
    mask = attention_mask.unsqueeze(-1).to(
        dtype=last_hidden_state.dtype
    )

    summed = (last_hidden_state * mask).sum(dim=1)

    counts = mask.sum(dim=1).clamp(min=1)

    return summed / counts


def cls_pooling(last_hidden_state):
    return last_hidden_state[:, 0, :]
```

```python
class TextClassifier(nn.Module):
    def __init__(self, pooling="mean", num_classes=4):
        super().__init__()

        if pooling not in ("mean", "cls"):
            raise ValueError("pooling must be 'mean' or 'cls'")

        self.pooling = pooling

        self.encoder = AutoModel.from_pretrained(MODEL_NAME)

        hidden_size = self.encoder.config.dim

        self.dropout = nn.Dropout(0.1)

        self.classifier = nn.Linear(
            hidden_size,
            num_classes
        )

    def forward(self, input_ids, attention_mask):
        outputs = self.encoder(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        hidden = outputs.last_hidden_state

        if self.pooling == "mean":
            pooled = mean_pooling(hidden, attention_mask)
        else:
            pooled = cls_pooling(hidden)

        pooled = self.dropout(pooled)

        logits = self.classifier(pooled)

        return logits
```

```python
torch.manual_seed(SEED)

model = TextClassifier(
    pooling="mean",
    num_classes=NUM_CLASSES
)

model = model.to(device)

print("Model device:", next(model.parameters()).device)

total_params = sum(
    p.numel() for p in model.parameters()
)

trainable_params = sum(
    p.numel() for p in model.parameters()
    if p.requires_grad
)

print(f"Total parameters: {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
```

```python

import gc
import time
import random
from pathlib import Path
from contextlib import nullcontext

from tqdm.auto import tqdm
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
from transformers import get_linear_schedule_with_warmup

SEED = 42
NUM_EPOCHS = 3
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01
MAX_GRAD_NORM = 1.0

CHECKPOINT_DIR = Path("checkpoints")
CHECKPOINT_DIR.mkdir(exist_ok=True)

# Освобождаем модель, созданную при проверке архитектуры
if "model" in globals():
    del model

gc.collect()

if device.type == "cuda":
    torch.cuda.empty_cache()
    USE_AMP = True
    AMP_DTYPE = (
        torch.bfloat16
        if torch.cuda.is_bf16_supported()
        else torch.float16
    )
else:
    USE_AMP = False
    AMP_DTYPE = torch.float32

USE_FP16_SCALER = (
    USE_AMP and AMP_DTYPE == torch.float16
)

def amp_context():
    if USE_AMP:
        return torch.autocast(
            device_type="cuda",
            dtype=AMP_DTYPE
        )
    return nullcontext()

print("Device:", device)
print("Mixed precision:", USE_AMP)
print("AMP dtype:", AMP_DTYPE)
print("Epochs:", NUM_EPOCHS)
```

```python
@torch.inference_mode()
def evaluate_model(model, data_loader):
    model.eval()

    criterion = torch.nn.CrossEntropyLoss()

    total_loss = 0.0
    total_examples = 0

    all_predictions = []
    all_labels = []

    for batch in tqdm(
        data_loader,
        desc="Validation",
        leave=False
    ):
        batch = {
            k: v.to(device, non_blocking=True)
            for k, v in batch.items()
        }

        with amp_context():
            logits = model(
                input_ids=batch["input_ids"],
                attention_mask=batch["attention_mask"]
            )

            loss = criterion(logits, batch["labels"])

        predictions = logits.argmax(dim=-1)

        batch_size = batch["labels"].size(0)

        total_loss += loss.item() * batch_size
        total_examples += batch_size

        all_predictions.extend(
            predictions.cpu().tolist()
        )
        all_labels.extend(
            batch["labels"].cpu().tolist()
        )

    return {
        "loss": total_loss / total_examples,
        "accuracy": accuracy_score(
            all_labels, all_predictions
        ),
        "precision": precision_score(
            all_labels,
            all_predictions,
            average="macro",
            zero_division=0
        ),
        "recall": recall_score(
            all_labels,
            all_predictions,
            average="macro",
            zero_division=0
        ),
        "f1": f1_score(
            all_labels,
            all_predictions,
            average="macro",
            zero_division=0
        )
    }
```

```python
def train_one_epoch(
    model,
    train_loader,
    optimizer,
    scheduler,
    scaler,
    epoch
):
    model.train()

    criterion = torch.nn.CrossEntropyLoss()

    total_loss = 0.0
    total_examples = 0

    progress_bar = tqdm(
        train_loader,
        desc=f"Epoch {epoch}/{NUM_EPOCHS}"
    )

    start_time = time.perf_counter()

    for step, batch in enumerate(progress_bar, start=1):

        batch = {
            k: v.to(device, non_blocking=True)
            for k, v in batch.items()
        }

        optimizer.zero_grad(set_to_none=True)

        with amp_context():
            logits = model(
                input_ids=batch["input_ids"],
                attention_mask=batch["attention_mask"]
            )

            loss = criterion(
                logits,
                batch["labels"]
            )

        scaler.scale(loss).backward()

        scaler.unscale_(optimizer)

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            MAX_GRAD_NORM
        )

        old_scale = scaler.get_scale()
        scaler.step(optimizer)
        scaler.update()

        if (
            not USE_FP16_SCALER
            or scaler.get_scale() >= old_scale
        ):
            scheduler.step()

        batch_size = batch["labels"].size(0)

        total_loss += loss.item() * batch_size
        total_examples += batch_size

        if step % 20 == 0:
            progress_bar.set_postfix(
                loss=f"{total_loss / total_examples:.4f}"
            )

        if epoch == 1 and step == 100:
            if device.type == "cuda":
                torch.cuda.synchronize()

            elapsed = time.perf_counter() - start_time

            print(
                f"\n100 batches in {elapsed:.1f} sec"
            )
            print(
                f"Average: {elapsed / 100:.3f} sec/batch"
            )

    if device.type == "cuda":
        torch.cuda.synchronize()

    epoch_time = time.perf_counter() - start_time

    return {
        "loss": total_loss / total_examples,
        "time": epoch_time
    }
```

```python
def run_experiment(pooling_type):
    print(f"\n{'=' * 55}")
    print(f"Starting experiment: {pooling_type}")
    print(f"{'=' * 55}")

    random.seed(SEED)
    np.random.seed(SEED)
    torch.manual_seed(SEED)

    if device.type == "cuda":
        torch.cuda.manual_seed_all(SEED)
        torch.cuda.reset_peak_memory_stats()

    model = TextClassifier(
        pooling=pooling_type,
        num_classes=NUM_CLASSES
    ).to(device)

    loader = make_train_loader(seed=SEED)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY
    )

    total_steps = len(loader) * NUM_EPOCHS
    warmup_steps = int(total_steps * 0.1)

    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=warmup_steps,
        num_training_steps=total_steps
    )

    scaler = torch.amp.GradScaler(
        "cuda",
        enabled=USE_FP16_SCALER
    )

    history = []
    best_f1 = -1.0
    best_metrics = None
    best_epoch = None

    checkpoint_path = (
        CHECKPOINT_DIR / f"distilbert_{pooling_type}.pt"
    )

    experiment_start = time.perf_counter()

    for epoch in range(1, NUM_EPOCHS + 1):

        train_stats = train_one_epoch(
            model,
            loader,
            optimizer,
            scheduler,
            scaler,
            epoch
        )

        val_stats = evaluate_model(
            model,
            val_loader
        )

        history.append({
            "epoch": epoch,
            "train_loss": train_stats["loss"],
            "val_loss": val_stats["loss"],
            "val_accuracy": val_stats["accuracy"],
            "val_precision": val_stats["precision"],
            "val_recall": val_stats["recall"],
            "val_f1": val_stats["f1"],
            "train_time_sec": train_stats["time"]
        })

        print(
            f"\nEpoch {epoch}: "
            f"train_loss={train_stats['loss']:.4f}, "
            f"val_loss={val_stats['loss']:.4f}, "
            f"val_f1={val_stats['f1']:.4f}"
        )

        if val_stats["f1"] > best_f1:
            best_f1 = val_stats["f1"]
            best_metrics = val_stats.copy()
            best_epoch = epoch

            torch.save({
                "model_state_dict": model.state_dict(),
                "epoch": epoch,
                "pooling": pooling_type,
                "val_metrics": val_stats
            }, checkpoint_path)

            print("Best model saved!")

    if device.type == "cuda":
        torch.cuda.synchronize()

    total_time = time.perf_counter() - experiment_start

    peak_memory_gib = (
        torch.cuda.max_memory_allocated() / 1024**3
        if device.type == "cuda"
        else None
    )

    history_df = pd.DataFrame(history)

    history_df.to_csv(
        CHECKPOINT_DIR / f"history_{pooling_type}.csv",
        index=False
    )

    print(f"\nBest epoch: {best_epoch}")
    print(f"Best validation F1: {best_f1:.4f}")
    print(f"Total time: {total_time / 60:.1f} minutes")

    if peak_memory_gib is not None:
        print(f"Peak VRAM: {peak_memory_gib:.2f} GiB")

    result = {
        "pooling": pooling_type,
        "history": history_df,
        "best_metrics": best_metrics,
        "best_epoch": best_epoch,
        "total_time": total_time,
        "peak_vram": peak_memory_gib,
        "checkpoint": str(checkpoint_path)
    }

    del model, optimizer, scheduler, scaler, loader
    gc.collect()

    if device.type == "cuda":
        torch.cuda.empty_cache()

    return result
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

@torch.inference_mode()
def test_checkpoint(pooling_type):
    model = TextClassifier(
        pooling=pooling_type,
        num_classes=NUM_CLASSES
    )

    checkpoint_path = (
        CHECKPOINT_DIR / f"distilbert_{pooling_type}.pt"
    )

    checkpoint = torch.load(
        checkpoint_path,
        map_location="cpu",
        weights_only=True
    )

    model.load_state_dict(
        checkpoint["model_state_dict"]
    )

    model = model.to(device)
    model.eval()

    y_true = []
    y_pred = []

    for batch in tqdm(
        test_loader,
        desc=f"Testing {pooling_type}"
    ):
        labels = batch["labels"].tolist()

        inputs = {
            k: v.to(device)
            for k, v in batch.items()
            if k != "labels"
        }

        with amp_context():
            logits = model(**inputs)

        predictions = logits.argmax(dim=-1)

        y_true.extend(labels)
        y_pred.extend(predictions.cpu().tolist())

    metrics = {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(
            y_true, y_pred, average="macro"
        ),
        "Recall": recall_score(
            y_true, y_pred, average="macro"
        ),
        "Macro F1": f1_score(
            y_true, y_pred, average="macro"
        )
    }

    del model
    gc.collect()

    if device.type == "cuda":
        torch.cuda.empty_cache()

    return {
        "metrics": metrics,
        "y_true": np.array(y_true),
        "y_pred": np.array(y_pred),
        "epoch": checkpoint["epoch"]
    }


mean_test = test_checkpoint("mean")
cls_test = test_checkpoint("cls")

print("Both models evaluated!")
```
